```python
# 하루치 사용량 시각화 함수
def plot_daily_usage(data, date):
    """
    Function: plot_daily_usage
        - 특정 날짜의 15분 단위 전력 사용량을 시각화하는 함수
    Parameters:
        - data: DataFrame, 'datetime' 열 포함
        - date: str, 'YYYY-MM-DD' 형식
    Returns:
        - None (그래프 출력)
    """
    # 1. 특정 날짜 데이터 필터링
    daily_data = data[data['datetime'].str.startswith(date)]
    
    # 2. datetime을 datetime 타입으로 변환
    daily_data['datetime'] = pd.to_datetime(daily_data['datetime'])
    
    # 3. 요일 계산 (예: 'Monday', 'Tuesday' 등)
    weekday = datetime.strptime(date, "%Y-%m-%d").strftime("%A") 
    # 4. 그래프
    plt.figure(figsize=(12, 6))
    plt.plot(daily_data['datetime'], daily_data['사용량 (kWh)'], marker='o')

    plt.title(f'{date} ({weekday})')
    plt.xlabel('Time')
    plt.ylabel('Usage (kWh)')
    plt.ylim(0, 40) # y축 범위 설정

    # 5. x축: 한 시간 간격으로만 표시
    plt.xticks(
        ticks=daily_data['datetime'][::4],  # 15분 × 4 = 1시간 간격
        labels=daily_data['datetime'][::4].dt.strftime('%H:%M'), # 시간:분 형식으로 표시
        rotation=45 # x축 레이블 45도 회전
    )

    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
# train 시작
plot_daily_usage(gm_gold_usage, '2024-08-06')
```

```python
# 요일별 평균 사용량 시각화
def plot_weekday_average_usage(data, date):
    """
    Function: plot_weekday_average_usage
        - 요일별 평균 전력 사용량을 시각화하는 함수
    Parameters:
        - data: DataFrame, must include 'datetime' and '사용량 (kWh)'
        - date: str, 'YYYY-MM' format, used to filter specific month
    Returns:
        - None (displays bar chart)
    """
    # 1. 데이터 복사 및 datetime 변환
    df = data.copy()
    df['datetime'] = pd.to_datetime(df['datetime'])
    
    # 2. 해당 월 데이터 필터링
    df = df[df['datetime'].dt.strftime('%Y-%m') == date]
    if df.empty:
        print(f"No data found for {date}.")
        return
    
    # 3. 요일별 평균 계산
    df['weekday'] = df['datetime'].dt.day_name()
    weekday_avg = df.groupby('weekday')['사용량 (kWh)'].mean().reindex(
        ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
    )
    
    # 4. 그래프
    plt.figure(figsize=(10, 5))
    plt.bar(weekday_avg.index, weekday_avg.values)
    
    plt.title(f'Average Daily Usage by Weekday ({date})')
    plt.xlabel('Weekday')
    plt.ylabel('Average Usage (kWh)')
    plt.ylim(0, 40)
    plt.tight_layout()
    plt.show()
```

```python
plot_weekday_average_usage(gm_gold_usage, "2024-08")
```
